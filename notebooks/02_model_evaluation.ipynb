```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
if not (ROOT / 'configs/dataset.yaml').is_file():
    raise RuntimeError('Open this notebook from the FruitVision root or notebooks directory.')
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
```

```python
import json
metrics_path = ROOT / 'outputs/metrics/baseline_test/metrics.json'
if not metrics_path.is_file():
    raise FileNotFoundError('Run python src/evaluate.py after training on your real dataset.')
report = json.loads(metrics_path.read_text())
print(report['metric_notes'])
report['overall']
```

```python
report['per_class']
```

```python
import matplotlib.pyplot as plt
values = report['overall']
fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(list(values), list(values.values()))
ax.set(ylim=(0, 1), ylabel='Score', title=f"Real evaluation: {report['run']['split']}")
fig.tight_layout()
fig.savefig(ROOT / 'outputs/figures/evaluation_summary.png', dpi=150)
plt.show()
```
